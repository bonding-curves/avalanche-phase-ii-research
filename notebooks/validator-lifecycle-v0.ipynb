# Validator lifecycle, v0: five origins against what happened

The first behaviour model. From an origin day, the validator commitments open at its end are the state. Each day, ending commitments renew on the same node with the observed seven-day probability, 80.15%, and draw a new duration, or leave. New commitments enter at the rate counted in the year before the origin, with durations and stakes drawn from that year. Every input precedes the origin, so the 90 days after it are a holdout.

**Status:** v0, unreviewed. Branch `prototype/model-v1`. The origins are the five network upgrades before the validation lockbox, which starts 2025-11-19 16:00 UTC. Each panel shows 20 runs as a pale pile of lines and what happened as the dark line.

```python
from datetime import date

import holoviews as hv
import hvplot.pandas  # noqa: F401
import pandas as pd

from avalanche_research import datasets
from avalanche_research.model.engine import run
from avalanche_research.model.lifecycle import ValidatorLifecycle, actual_path, origin_inputs, score
from avalanche_research.model.replay import daily_events
from avalanche_research.palette import QUANTITY_COLORS

hv.extension("bokeh")
RUN = "staking-periods-2026-09-30"
periods = str(datasets.parquet_path("derived.staking-periods", RUN))
actual = actual_path(daily_events(periods))
ORIGINS = {"Banff": date(2022, 10, 18), "Cortina": date(2023, 4, 25), "Durango": date(2024, 3, 6), "Etna": date(2024, 12, 16), "Fortuna": date(2025, 4, 8)}
```

```python
tables, scores = {}, []
for name, origin in ORIGINS.items():
    tables[name] = run(ValidatorLifecycle(origin=origin, inputs_source=RUN, **origin_inputs(periods, origin)))
    scores.append(score(tables[name], actual, origin).assign(origin=name))
scores = pd.concat(scores)
```

## Active validators after each origin

```python
def panel(name, variable, scale, color, label):
    origin = ORIGINS[name]
    runs = tables[name][tables[name]["variable"] == variable].assign(value=lambda t: t["value"] / scale)
    seen = actual[(actual["date_utc"] >= origin) & (actual["date_utc"] <= origin + pd.Timedelta(days=90))]
    seen = seen.assign(timestep=range(len(seen)), value=seen[variable] / scale)
    pile = runs.hvplot.line(x="timestep", y="value", by="run", color=color, alpha=0.25, line_width=1, legend=False)
    return (pile * seen.hvplot.line(x="timestep", y="value", color="#222222", line_width=2.5)).opts(
        title=f"{name}, {origin}", xlabel="Days after the origin", ylabel=label, width=420, height=260, show_legend=False)

hv.Layout([panel(n, "active_validators", 1, QUANTITY_COLORS["nodes"], "Active validators") for n in ORIGINS]).cols(3)
```

## Validator stake after each origin

```python
hv.Layout([panel(n, "active_validator_stake_navax", 1e15, QUANTITY_COLORS["stake"], "Validator stake, million AVAX") for n in ORIGINS]).cols(3)
```

## Scores

Mean error of the 20 runs against the actual value, beside the error of persistence, which holds the origin value. The model wins a row when its error is smaller. In v0 it wins 8 of 20 rows, so it is no better than persistence. Its 10 to 90 percent band covers the actual stake in 7 of 10 rows but the actual validator count in only 1 of 10: the count moves more than one renewal rate and one entry rate allow. That is the baseline fitted blocks must beat: renewal by tenure, named operators such as GoGoPool and BENQI as events, and duration choice weighted by stake.

```python
view = scores.assign(
    model_wins=scores["model_abs_error"] < scores["persistence_abs_error"],
    band_covers=(scores["model_p10"] <= scores["actual"]) & (scores["actual"] <= scores["model_p90"]),
)
view.pivot_table(index=["origin", "horizon_days"], columns="variable", values=["model_wins", "band_covers"], aggfunc="first")
```
